## Problem 1: Build a Financial Summary Line

Given `company`, `revenue`, and `expenses`, create a single f-string that prints revenue, expenses, profit, and margin.

**Requirements**

- Use only one f-string.
- Calculate profit inside the f-string.
- Calculate margin inside the f-string.
- Format money values with commas and two decimal places.
- Format the margin as a percentage with two decimal places.

````python
company = 'DataForge'
revenue = 1_257_430.75
expenses = 983_271.40

summary = f'{company} | Revenue: ${revenue:,.2f} | Expenses: ${expenses:,.2f} | Profit: ${revenue - expenses:,.2f} | Margin: {(revenue - expenses) / revenue:.2%}'

summary
````

### Solution Notes

`:,.2f` adds thousands separators and keeps two decimal places. `:.2%` formats a ratio as a percentage.

## Problem 2: Right-Align a Table of Scores

Print a table of students and scores using f-string alignment.

**Requirements**

- Left-align names in a width of 12.
- Right-align scores.
- Show exactly two digits after the decimal point.

````python
students = [
    ('Alice', 91.236),
    ('Brandon', 84.5),
    ('Christopher', 99.995),
    ('Diana', 73.1),
]

print(f'{"Student":<12} {"Score":>8}')
print('-' * 22)

for name, score in students:
    print(f'{name:<12} {score:>7.2f}%')
````

### Solution Notes

`:<12` means left-align in a field of width 12. `:>7.2f` means right-align in a field of width 7 and show two decimals.

## Problem 5: Use f-Strings with Dictionaries

Access dictionary values inside an f-string and use a conditional expression for the status.

````python
user = {
    'first_name': 'Ada',
    'last_name': 'Lovelace',
    'role': 'Mathematician',
    'active': True,
}

line = f"{user['first_name']} {user['last_name']} | Role: {user['role']} | Status: {'ACTIVE' if user['active'] else 'INACTIVE'}"

line
````

## Problem 6: Use f-Strings with Object Attributes

Create a `Product` class and use an f-string to calculate and format a receipt line.

````python
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity


item = Product('Keyboard', 79.99, 3)

receipt_line = f'{item.name} x {item.quantity} = ${item.price * item.quantity:.2f}'

receipt_line
````

### Solution Notes

f-strings can access object attributes and evaluate arithmetic expressions directly.

## Problem 7: Format Binary, Hexadecimal, and Decimal Values

Use f-string format specifiers to display one integer in multiple bases.

````python
number = 255

formatted = f'decimal={number:d} | binary={number:b} | hex={number:x} | HEX={number:X}'

formatted
````

Format a `datetime` object directly inside an f-string.

## Problem 9: Use f-Strings Inside a Closure

Write `make_labeler(prefix)` so it returns a function that labels messages.

````python
def make_labeler(prefix):
    def label(value):
        return f'[{prefix}] {value}'
    return label


error_label = make_labeler('ERROR')

error_label('File not found')
````

### Solution Notes

The variable `prefix` is captured by the inner function as a free variable, even though it is only referenced inside the f-string.

## Problem 10: Avoid Overly Complex f-Strings

Refactor a complex f-string by moving business logic outside the string.

````python
orders = [120, 80, 50, 200]
discount = 0.10

eligible_orders = [amount for amount in orders if amount >= 100]
discounted_total = sum(amount * (1 - discount) for amount in eligible_orders)

result = f'Total after discount: ${discounted_total:.2f}'

result
````

### Solution Notes

Best practice: use f-strings for presentation, not for hiding complicated logic.

## Problem 12: Implement `__format__` for Custom f-String Formatting

Create a `Money` class that supports custom formatting in f-strings.

````python
class Money:
    def __init__(self, amount):
        self.amount = amount

    def __format__(self, format_spec):
        if not format_spec:
            format_spec = ',.2f'
        return f'${self.amount:{format_spec}}'


price = Money(1234.5)

print(f'{price}')
print(f'{price:.1f}')
````

### Solution Notes

When an object appears inside an f-string, Python can call its `__format__` method.

## Problem 13: Understand Conversion Flags

Use `{value}` and `{value!r}` to compare `str()` and `repr()` behavior.

````python
class Message:
    def __str__(self):
        return 'Hello\nWorld'

    def __repr__(self):
        return 'Message("Hello\\nWorld")'


value = Message()

print(f'Normal string form: {value}')
print(f'Representation: {value!r}')
````

### Solution Notes

`{value}` uses `str(value)` by default. `{value!r}` uses `repr(value)`, which is often better for debugging.

## Problem 14: Fix a Python 3.6 f-String Syntax Problem

The following is invalid in Python 3.6:

```python
name = 'Python'
result = f'{name\n}'
```

Rewrite it correctly so the newline is outside the expression.

````python
name = 'Python'

result = f'{name}\n'

print(result)
````

### Solution Notes

In Python 3.6, backslashes are not allowed inside the expression part of an f-string. Put the newline outside `{...}`.

# Final Best Practices
