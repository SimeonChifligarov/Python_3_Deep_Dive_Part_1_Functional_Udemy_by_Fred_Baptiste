## Problem 1: Build a Financial Summary Line

You are given the following values:

```python
company = 'DataForge'
revenue = 1_257_430.75
expenses = 983_271.40
```

Create a single f-string that produces this exact style of output:

```text
DataForge | Revenue: $1,257,430.75 | Expenses: $983,271.40 | Profit: $274,159.35 | Margin: 21.80%
```

### Requirements

- Use only one f-string for the final output.
- Calculate profit inside the f-string.
- Calculate margin inside the f-string.
- Format all money values with commas and two decimal places.
- Format the margin as a percentage with two decimal places.


````python
company = 'DataForge'
revenue = 1_257_430.75
expenses = 983_271.40

summary = f'{company} | Revenue: ${revenue:,.2f} | Expenses: ${expenses:,.2f} | Profit: ${revenue - expenses:,.2f} | Margin: {(revenue - expenses) / revenue:.2%}'

summary
````

### Explanation

The format specifier `:,.2f` means:

- `,` adds thousands separators
- `.2f` keeps exactly two digits after the decimal point

The format specifier `:.2%` means:

- multiply the value by 100
- display it as a percentage
- keep two digits after the decimal point

The expression:

```python
(revenue - expenses) / revenue
```

is evaluated directly inside the f-string.

### Best Practice

This solution is acceptable because the calculations are short and obvious.

If the calculation became longer or was reused, it would be better to compute `profit` and `margin` before the f-string.


## Problem 2: Right-Align a Table of Scores

Given this data:

```python
students = [
    ('Alice', 91.236),
    ('Brandon', 84.5),
    ('Christopher', 99.995),
    ('Diana', 73.1)
]
```

Print a table that looks like this:

```text
Student          Score
----------------------
Alice           91.24%
Brandon         84.50%
Christopher    100.00%
Diana           73.10%
```

### Requirements

- Use f-strings for the header and each row.
- Left-align names in a width of 12.
- Right-align scores.
- Format scores with exactly two decimal places followed by `%`.


````python
students = [
    ('Alice', 91.236),
    ('Brandon', 84.5),
    ('Christopher', 99.995),
    ('Diana', 73.1),
]

print(f'{"Student":<12} {"Score":>8}')
print('-' * 22)

for name, score in students:
    print(f'{name:<12} {score:>7.2f}%')
````

### Explanation

The format specifier `:<12` means:

```text
left-align the value in a field of width 12
```

The format specifier `:>7.2f` means:

```text
right-align the value in a field of width 7 and show two decimal places
```

The percent sign is added manually because the input values are already percentages such as `91.236`, not ratios such as `0.91236`.

### Best Practice

Use alignment specifiers when creating console reports, logs, or fixed-width text output.

For larger tabular data, prefer libraries such as `tabulate`, `pandas`, or CSV output.


## Problem 3: Dynamic Width and Precision

You are given:

```python
value = 3.141592653589793
width = 12
precision = 4
```

Use an f-string to format `value` so that:

- the total field width is controlled by `width`
- the number of decimal places is controlled by `precision`
- the result is right-aligned

Expected output:

```text
      3.1416
```


````python
value = 3.141592653589793
width = 12
precision = 4

formatted = f'{value:>{width}.{precision}f}'

formatted
````

### Explanation

Format specifications can contain expressions.

In this example:

```python
f'{value:>{width}.{precision}f}'
```

Dynamic formatting is useful when output rules are configurable.

However, avoid hiding complicated layout rules inside a single f-string. If formatting rules become complex, assign intermediate names first.


## Problem 4: Escape Literal Braces

Create an f-string that produces this exact output:

```text
The configuration is: {host: 'localhost', port: 8080}
```

Given:

```python
host = 'localhost'
port = 8080
```

### Requirements

- Use an f-string.
- The outer braces must appear literally in the final string.
- The values of `host` and `port` must be interpolated.


````python
host = 'localhost'
port = 8080

result = f"The configuration is: {{host: '{host}', port: {port}}}"

result
````

### Explanation

Inside an f-string, literal braces must be doubled.

Use:

```python
{{ 
```

to output:

```text
{
```

Use:

```python
}}
```

to output:

```text
}
```

Single braces are reserved for Python expressions.

### Best Practice

When a string needs many literal braces, consider whether a dictionary, JSON serializer, or template engine would be clearer.


## Problem 5: Use f-Strings with Dictionaries

You are given a dictionary:

```python
user = {
    'first_name': 'Ada',
    'last_name': 'Lovelace',
    'role': 'Mathematician',
    'active': True
}
```

Create this exact output using an f-string:

```text
Ada Lovelace | Role: Mathematician | Status: ACTIVE
```

### Requirements

- Access dictionary values directly inside the f-string.
- Use a conditional expression inside the f-string.
- Convert `True` into `ACTIVE`.
- Convert `False` into `INACTIVE`.


````python
user = {
    'first_name': 'Ada',
    'last_name': 'Lovelace',
    'role': 'Mathematician',
    'active': True,
}

line = f"{user['first_name']} {user['last_name']} | Role: {user['role']} | Status: {'ACTIVE' if user['active'] else 'INACTIVE'}"

line
````

### Explanation

Dictionary lookups can be used inside f-string expressions.

Because the outer string uses double quotes, the dictionary keys can safely use single quotes:

```python
f"{user['first_name']}"
```

The expression:

```python
'ACTIVE' if user['active'] else 'INACTIVE'
```

is a conditional expression.

### Best Practice

This is okay for short dictionary lookups.

If the f-string contains many repeated dictionary accesses, consider assigning local variables first:

```python
first_name = user['first_name']
last_name = user['last_name']
```


## Problem 6: Use f-Strings with Object Attributes

Create a class called `Product` with attributes:

- `name`
- `price`
- `quantity`

Then create a product:

```python
Product('Keyboard', 79.99, 3)
```

Use an f-string to produce:

```text
Keyboard x 3 = $239.97
```

### Requirements

- Access object attributes inside the f-string.
- Calculate the total inside the f-string.
- Format the total as dollars with two decimal places.


````python
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity


item = Product('Keyboard', 79.99, 3)

receipt_line = f'{item.name} x {item.quantity} = ${item.price * item.quantity:.2f}'

receipt_line
````

### Explanation

f-strings can access object attributes directly:

```python
f'{item.name}'
```

They can also evaluate arithmetic expressions:

```python
f'{item.price * item.quantity:.2f}'
```

### Best Practice

For real applications, consider making `total` a property:

```python
@property
def total(self):
    return self.price * self.quantity
```

Then the f-string becomes simpler:

```python
f'{item.name} x {item.quantity} = ${item.total:.2f}'
```

## Problem 9: Use f-Strings Inside a Closure

Write a function called `make_labeler(prefix)`.

It should return an inner function called `label(value)`.

The returned function should produce strings like this:

```python
error_label = make_labeler('ERROR')
error_label('File not found')
```

Expected output:

```text
[ERROR] File not found
```

### Requirements

- Use a closure.
- Use an f-string inside the inner function.
- The inner function must reference `prefix` only inside the f-string.


````python
def make_labeler(prefix):
    def label(value):
        return f'[{prefix}] {value}'
    return label


error_label = make_labeler('ERROR')

error_label('File not found')
````

### Explanation

The variable `prefix` is a free variable captured by the inner function.

Python correctly recognizes it even when it is only referenced inside the f-string.

### Best Practice

Closures are useful for creating specialized formatting functions.

However, for complex labeling behavior, a class may be more readable.


## Problem 10: Avoid Overly Complex f-Strings

The following f-string works, but it is difficult to read:

```python
orders = [120, 80, 50, 200]
discount = 0.10

result = f'Total after discount: ${sum(x - x * discount for x in orders if x >= 100):.2f}'
```

Refactor this code using best practices.

### Requirements

- Move complex logic outside the f-string.
- Keep the final f-string simple.
- Produce the same output.


````python
orders = [120, 80, 50, 200]
discount = 0.10

eligible_orders = [amount for amount in orders if amount >= 100]
discounted_total = sum(amount * (1 - discount) for amount in eligible_orders)

result = f'Total after discount: ${discounted_total:.2f}'

result
````

### Explanation

Although f-strings can contain complex expressions, they should usually stay readable.

A good rule is:

```text
Use f-strings for presentation, not for hiding business logic.
```

The refactored version is easier to debug, test, and maintain.

### Best Practice

When an f-string contains a generator expression, nested conditionals, or multiple operations, consider computing a variable first.

## Problem 12: Implement `__format__` for Custom f-String Formatting

Create a class called `Money` that stores an amount.

It should support this behavior:

```python
price = Money(1234.5)

f'{price}'
```

Expected output:

```text
$1,234.50
```

And this:

```python
f'{price:.1f}'
```

Expected output:

```text
$1,234.5
```

### Requirements

- Implement `__format__`.
- If no format specification is provided, use `,.2f`.
- If a format specification is provided, use it.


````python
class Money:
    def __init__(self, amount):
        self.amount = amount

    def __format__(self, format_spec):
        if not format_spec:
            format_spec = ',.2f'
        return f'${self.amount:{format_spec}}'


price = Money(1234.5)

print(f'{price}')
print(f'{price:.1f}')
````

### Explanation

f-strings call the formatting protocol behind the scenes.

When Python evaluates:

```python
f'{price:.1f}'
```

it calls:

```python
price.__format__('.1f')
```

This allows custom classes to define how they should appear inside formatted strings.

### Best Practice

Use `__format__` when your object has a natural display format that may need variations.


## Problem 13: Understand Conversion Flags

Given this class:

```python
class Message:
    def __str__(self):
        return 'Hello\nWorld'

    def __repr__(self):
        return 'Message("Hello\\nWorld")'
```

Use f-strings to display:

1. the normal string form
2. the developer-friendly representation

### Requirements

- Use `{value}` for the normal string form.
- Use `{value!r}` for the representation.


````python
class Message:
    def __str__(self):
        return 'Hello\nWorld'

    def __repr__(self):
        return 'Message("Hello\\nWorld")'


value = Message()

print(f'Normal string form: {value}')
print(f'Representation: {value!r}')
````

### Explanation

`{value}` uses `str(value)` by default.

`{value!r}` uses `repr(value)`.

This is useful when debugging because `repr` often shows escape characters and structural details more clearly.

### Best Practice

Use `!r` in debugging messages and logs when you need an unambiguous representation.


## Problem 14: Fix a Python 3.6 f-String Syntax Problem

The following code is invalid in Python 3.6:

```python
name = 'Python'
result = f'{name\n}'
```

Explain why it is invalid and rewrite it correctly so the result is:

```text
Python
```

followed by a newline character.


````python
name = 'Python'

result = f'{name}\n'

print(result)
````

### Explanation

In Python 3.6, backslashes are not allowed inside the expression part of an f-string.

This is invalid:

```python
f'{name\n}'
```

because the backslash appears inside `{...}`.

The correct solution is to put the newline outside the expression:

```python
f'{name}\n'
```

### Best Practice

Keep escape sequences outside f-string expression braces.


## Problem 15: Generate a Safe SQL Preview String

You are given:

```python
table = 'users'
limit = 10
```

Use an f-string to generate this preview string:

```text
SELECT * FROM users LIMIT 10;
```

Then explain why this technique should not be used to insert untrusted user input into SQL queries.


````python
table = 'users'
limit = 10

query_preview = f'SELECT * FROM {table} LIMIT {limit};'

query_preview
````

### Explanation

f-strings are fine for trusted display strings or controlled internal values.

However, they should not be used to insert untrusted user input directly into SQL queries.

Unsafe example:

```python

```

A malicious user could provide input that changes the meaning of the query.

### Best Practice

## Problem 19: Format a Ratio Only When Denominator Is Nonzero

## Problem 20: Build a Reusable Report Row Function

Write a function called `format_row(name, value, width=15, precision=2)`.

It should return a string like:

```text
revenue        $1234.57
```

Given:

```python
format_row('revenue', 1234.567)
```

### Requirements

- Use an f-string.
- Use dynamic width.
- Use dynamic precision.
- Format the value as currency.


````python
def format_row(name, value, width=15, precision=2):
    return f'{name:<{width}} ${value:.{precision}f}'


format_row('revenue', 1234.567)
````

### Explanation

The expression `{width}` is used inside the format specification:

```python
{name:<{width}}
```

The expression `{precision}` controls the number of decimal places:

```python
{value:.{precision}f}
```

### Best Practice

This is a good use of f-strings because the formatting logic is reusable and named.


# Final Best Practices for f-Strings

Use f-strings when:

- you need readable interpolation
- you want concise formatting
- you are formatting values for display
- expressions are simple and obvious
- you want to use Python's format mini-language

Avoid f-strings when:

- the expression becomes too complex
- the string contains untrusted input for SQL, shell commands, or code execution
- older Python versions before 3.6 must be supported
- a dedicated serializer would be safer, such as JSON output

General rule:

```text
Keep calculations readable. Keep f-strings focused on formatting.
```

Another useful rule:

```text
If you need to pause to understand the expression inside `{...}`, move that expression to a named variable.
```
