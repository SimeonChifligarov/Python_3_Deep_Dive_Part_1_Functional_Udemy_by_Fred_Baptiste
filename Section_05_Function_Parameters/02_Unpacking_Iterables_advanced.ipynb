## Problem 1: Tuple or Not?

Predict the type of each variable before running the code.

````python
values = {
    'a': (10),
    'b': (10,),
    'c': 10,
    'd': (),
    'e': tuple(),
    'f': 10,  # note: this is not inside the dictionary as a tuple expression
}

for name, value in values.items():
    print(name, value, type(value))
````

### Solution 1

`(10)` is just the integer `10`. Parentheses only group expressions.

`(10,)` is a tuple because the comma creates the tuple.

`()` and `tuple()` both create empty tuples.

Best practice: when creating a one-item tuple, always include the trailing comma.

## Problem 2: Exact-Length Unpacking

Which assignments succeed and which fail?

````python
tests = [
    "a, b, c = [1, 2, 3]",
    "a, b = [1, 2, 3]",
    "a, b, c, d = [1, 2, 3]",
    "a, b, c = 'XYZ'",
    "a, b = 'XYZ'"
]

for test in tests:
    try:
        namespace = {}
        exec(test, {}, namespace)
        print('VALID:  ', test, '=>', namespace)
    except Exception as e:
        print('INVALID:', test, '=>', type(e).__name__, e)
````

### Solution 2

Normal unpacking requires the number of variables to exactly match the number of values.

Valid:

```python
a, b, c = [1, 2, 3]
a, b, c = 'XYZ'
```

Invalid:

```python
a, b = [1, 2, 3]
a, b, c, d = [1, 2, 3]
a, b = 'XYZ'
```

Too many or too few values cause a `ValueError`.

## Problem 3: Swap Without Temporary Variables

Given `x = 100` and `y = 200`, swap their values using unpacking.

````python
x = 100
y = 200

print('Before:', x, y)

x, y = y, x

print('After: ', x, y)
````

### Solution 3

Expected output:

```text
Before: 100 200
After:  200 100
```

Python evaluates the right-hand side first, then assigns the resulting values to the left-hand side.

## Problem 4: Rotate Three Variables

Use unpacking to rotate the values so that:

- `a` gets the old value of `c`
- `b` gets the old value of `a`
- `c` gets the old value of `b`

````python
a, b, c = 'A', 'B', 'C'

print('Before:', a, b, c)

a, b, c = c, a, b

print('After: ', a, b, c)
````

### Solution 4

Expected output:

```text
Before: A B C
After:  C A B
```

This is a clean use of unpacking for simultaneous reassignment.

## Problem 5: Unpack Records

Given a list of user records, unpack each record into meaningful variable names and print a sentence for each user.

````python
users = [
    ('Ada', 'Lovelace', 36),
    ('Grace', 'Hopper', 85),
    ('Guido', 'van Rossum', 69)
]

for first, last, age in users:
    print(f'{first} {last} is {age} years old.')
````

### Solution 5

This is more readable than indexing:

```python
record[0], record[1], record[2]
```

## Problem 6: Nested Unpacking

Unpack the nested data structure below into separate variables.

````python
record = ('Alice', ('Engineering', 'Backend'), (2020, 2024))

name, (department, team), (start_year, current_year) = record

print(name)
print(department)
print(team)
print(start_year)
print(current_year)
````

### Solution 6

Nested unpacking mirrors the structure of the data.

```python
name, (department, team), (start_year, current_year) = record
```

This works because the left-hand side has the same shape as the right-hand side.

## Problem 7: Dictionary Unpacking Trap

Predict what gets unpacked from this dictionary.

````python
person = {
    'name': 'Maya',
    'role': 'developer',
        "level": "senior"
}

a, b, c = person

print(a)
print(b)
print(c)
````

### Solution 7

Unpacking a dictionary directly unpacks its keys, not its values.

Expected output:

```text
name
role
level
```

To unpack values, use:

```python
name, role, level = person.values()
```

To unpack key-value pairs, use:

```python
for key, value in person.items():
    ...
```

## Problem 8: Safely Iterate Through Dictionary Items

Print each key and value from the dictionary using unpacking.

````python
config = {
    'host': 'localhost',
    'port': 5432,
    'debug': True
}

for key, value in config.items():
    print(f'{key} => {value}')
````

### Solution 8

`dict.items()` produces pairs like:

```python
('host', 'localhost')
('port', 5432)
('debug', True)
```

Each pair can be unpacked into `key` and `value`.

## Problem 9: Set Unpacking and Order Risk

Explain why the following code is risky.

````python
letters = {'p', 'y', 't', 'h', 'o', 'n'}

a, b, c, d, e, f = letters

print(a, b, c, d, e, f)
````

### Solution 9

The code may run, but it should not rely on a meaningful order.

Sets are unordered collections. You can unpack them only when you care about the number of elements, not their order.

Safer alternative:

```python
a, b, c, d, e, f = sorted(letters)
```

Use `sorted()` when deterministic order matters.

## Problem 10: Extended Unpacking

Use extended unpacking to separate the first item, the last item, and everything in the middle.

````python
numbers = [10, 20, 30, 40, 50, 60]

first, *middle, last = numbers

print('first:', first)
print('middle:', middle)
print('last:', last)
````

```text
first: 10
middle: [20, 30, 40, 50]
last: 60
```

The starred variable collects the remaining values into a list.

## Problem 11: Ignore Unneeded Values

Given a record with five fields, unpack only the name and email. Ignore the rest.

````python
record = ('Mina', 'mina@example.com', 'admin', True, '2024-01-15')

name, email, *_ = record

print(name)
print(email)
````

This is a common Python convention.

## Problem 12: Validate Shape Before Unpacking

Write a function that accepts a coordinate pair and returns a formatted string. It should reject invalid inputs cleanly.

````python
def format_point(point):
    try:
        x, y = point
    except ValueError:
        raise ValueError('point must contain exactly two values')

    return f'x={x}, y={y}'

print(format_point((10, 20)))
print(format_point([3, 4]))

try:
    print(format_point((1, 2, 3)))
except ValueError as e:
    print(type(e).__name__, e)
````

### Solution 12

The function expects exactly two values.

Valid inputs:

```python
(10, 20)
[3, 4]
```

Invalid input:

```python
(1, 2, 3)
```

Unpacking is useful, but it should be guarded when input shape is uncertain.

## Challenge Problem: Parse Log Entries

Each log entry has this structure:

```python
(timestamp, level, service, message, *tags)
```

Write a function that processes a list of log entries and returns dictionaries with named fields.

````python
logs = [
    ('2024-01-01 10:00:00', 'INFO', 'auth', 'User logged in', 'user:42'),
    ('2024-01-01 10:01:00', 'ERROR', 'payments', 'Payment failed', 'order:99', 'retry:true'),
    ('2024-01-01 10:02:00', 'WARNING', 'api', 'Slow response')
]

def parse_logs(entries):
    parsed = []

    for timestamp, level, service, message, *tags in entries:
        parsed.append({
            'timestamp': timestamp,
            'level': level,
            'service': service,
            'message': message,
            'tags': tags
        })

    return parsed

for entry in parse_logs(logs):
    print(entry)
````

### Challenge Solution

The first four fields are required and unpacked normally:

```python
timestamp, level, service, message
```

The remaining fields are optional and collected by:

```python
*tags
```

This makes the parser flexible while keeping the required structure clear.
