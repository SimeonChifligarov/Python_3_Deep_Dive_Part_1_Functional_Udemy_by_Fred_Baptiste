## Problem 1: Minimum One Argument Average

Write a function `avg(first, *rest)` that computes the average of one or more numbers.

The function should not allow zero arguments.

````python
def avg(first, *rest):
    values = (first, *rest)
    return sum(values) / len(values)


print(avg(10))
print(avg(10, 20, 30, 40))
````

### Solution explanation

`first` makes at least one positional argument mandatory.

`*rest` collects any remaining positional arguments into a tuple.

This avoids the `ZeroDivisionError` problem that can happen with `def avg(*args)` when no arguments are passed.

## Problem 2: Product of Arbitrary Arguments

Write a function `product(*args)` that returns the product of all positional arguments.

If no arguments are supplied, return `1`.

````python
def product(*args):
    result = 1
    for value in args:
        result *= value
    return result


print(product())
print(product(2, 3, 4))
print(product(10, -2, 5))
````

### Solution explanation

`*args` allows the function to receive any number of positional arguments.

The identity value for multiplication is `1`, so returning `1` for no arguments is mathematically consistent.

## Problem 3: Separate Required and Optional Scores

Write a function `score_report(name, first_score, *other_scores)` that returns a dictionary containing:

- the student's name
- all scores as a tuple
- the best score
- the worst score
- the average score

````python
def score_report(name, first_score, *other_scores):
    scores = (first_score, *other_scores)
    return {
        'name': name,
        'scores': scores,
        'best': max(scores),
        'worst': min(scores),
        'average': sum(scores) / len(scores)
    }


print(score_report('Ada', 95, 87, 100, 91))
````

### Solution explanation

`name` and `first_score` are required positional arguments.

`*other_scores` collects all additional scores.

Using a required `first_score` guarantees that there is at least one score to summarize.

## Problem 4: Forward Positional Arguments

Write a function `debug_call(func, *args)` that:

1. prints the function name
2. prints the positional arguments
3. calls the function with those arguments
4. returns the result

````python
def debug_call(func, *args):
    print('Calling:', func.__name__)
    print('Arguments:', args)
    return func(*args)


def add(a, b, c):
    return a + b + c


result = debug_call(add, 10, 20, 30)
print('Result:', result)
````

### Solution explanation

`debug_call` receives a function and any number of positional arguments.

`func(*args)` unpacks the collected tuple back into positional arguments when calling `func`.

## Problem 5: Validate Argument Count Manually

Write a function `exactly_three(*args)` that accepts exactly three positional arguments.

If the caller provides anything other than three arguments, raise a `TypeError` with a clear message.

````python
def exactly_three(*args):
    if len(args) != 3:
        raise TypeError(f'exactly_three expected 3 arguments, got {len(args)}')
    a, b, c = args
    return a, b, c


print(exactly_three(1, 2, 3))

try:
    exactly_three(1, 2)
except TypeError as ex:
    print(ex)
````

### Solution explanation

`*args` accepts any number of positional arguments, so the function must manually validate the length.

After validation, normal unpacking can safely assign the three values.

## Problem 6: Positional Argument Unpacking from Lists

Given a function:

```python
def rectangle_area(width, height):
    return width * height
```

Call it using a list `dimensions = [12, 5]`.

````python
def rectangle_area(width, height):
    return width * height


dimensions = [12, 5]

print(rectangle_area(*dimensions))
````

### Solution explanation

`rectangle_area(dimensions)` would pass the entire list as one argument.

`rectangle_area(*dimensions)` unpacks the list so that `12` becomes `width` and `5` becomes `height`.

## Problem 7: Combine Fixed and Unpacked Arguments

Given:

```python
def build_path(root, *parts):
    ...
```

Use `*args` to build a slash-separated path.

````python
def build_path(root, *parts):
    return '/'.join((root, *parts))


print(build_path('home', 'user', 'projects', 'python'))

extra_parts = ['data', 'raw', 'input.csv']
print(build_path('mnt', *extra_parts))
````

### Solution explanation

`root` is required.

`*parts` collects any remaining path components.

When calling the function, `*extra_parts` unpacks the list into separate positional arguments.

## Problem 8: Reject Non-Numeric Arguments

Write a function `numeric_sum(*args)` that sums arbitrary positional arguments, but raises `TypeError` if any argument is not an `int` or `float`.

````python
def numeric_sum(*args):
    for value in args:
        if not isinstance(value, (int, float)):
            raise TypeError(f'Expected int or float, got {type(value).__name__}')
    return sum(args)


print(numeric_sum(1, 2.5, 3))

try:
    numeric_sum(1, '2', 3)
except TypeError as ex:
    print(ex)
````

### Solution explanation

`*args` makes the function flexible, but that flexibility means validation is often needed.

The function loops over each argument and checks its type before calling `sum`.

## Problem 9: Difference Between Collected and Unpacked Arguments

Explain the difference between these two calls:

```python
func(values)
func(*values)
```

Use code to demonstrate the difference.

````python
def show_args(*args):
    print(args)


values = [10, 20, 30]

show_args(values)
show_args(*values)
````

### Solution explanation

`show_args(values)` passes the list as one positional argument.

`show_args(*values)` unpacks the list and passes each element as a separate positional argument.

So the first call produces a tuple containing one list, while the second produces a tuple containing three integers.

## Problem 10: Wrapper Around `max`

Write a function `loud_max(*args)` that prints all received arguments and then returns the maximum value.

````python
def loud_max(*args):
    print('Received:', args)
    return max(args)


print(loud_max(4, 9, 2, 11, 3))
````

### Solution explanation

`*args` captures all positional values.

Since `max` can accept an iterable, we can pass the collected `args` tuple directly to `max(args)`.

## Problem 11: Forward Arguments to Multiple Functions

Write a function `apply_all(functions, *args)` that calls every function in `functions` with the same positional arguments and returns a list of results.

````python
def apply_all(functions, *args):
    return [func(*args) for func in functions]


def add(a, b):
    return a + b


def multiply(a, b):
    return a * b


def power(a, b):
    return a ** b


functions = [add, multiply, power]
print(apply_all(functions, 2, 5))
````

### Solution explanation

`apply_all` accepts arbitrary positional arguments once and forwards them to each function using `func(*args)`.

This pattern is common in wrappers, decorators, callbacks, and plugin systems.

## Problem 12: Keyword-Only Argument After `*args`

Write a function `tagged_sum(*args, label)` that requires `label` to be passed as a keyword-only argument.

Return a string like:

```python
'total=15'
```

````python
def tagged_sum(*args, label):
    return f'{label}={sum(args)}'


print(tagged_sum(1, 2, 3, 4, 5, label='total'))

try:
    tagged_sum(1, 2, 3, 'total')
except TypeError as ex:
    print(ex)
````

### Solution explanation

Any parameter after `*args` becomes keyword-only.

Therefore, `label` cannot be supplied positionally. It must be supplied as `label='total'`.

## Problem 13: Enforce a Separator Keyword

Write a function `join_items(*items, sep=', ')` that joins any number of items into a string using a keyword-only separator.

````python
def join_items(*items, sep=', '):
    return sep.join(str(item) for item in items)


print(join_items('red', 'green', 'blue'))
print(join_items('red', 'green', 'blue', sep=' | '))
````

### Solution explanation

`*items` captures all positional arguments.

`sep` appears after `*items`, so it is keyword-only.

This avoids confusion between data items and configuration options.

## Problem 14: Argument Forwarding with Transformation

Write a function `call_with_doubled_args(func, *args)` that doubles every positional argument before calling `func`.

````python
def call_with_doubled_args(func, *args):
    doubled = [arg * 2 for arg in args]
    return func(*doubled)


def combine(a, b, c):
    return a + b + c


print(call_with_doubled_args(combine, 1, 2, 3))
````

### Solution explanation

`args` is a tuple of received positional arguments.

The function creates a transformed list called `doubled`.

`func(*doubled)` unpacks that list into the function call.

## Problem 15: Common Mistake with Lists

Why does this fail?

```python
def add_three(a, b, c):
    return a + b + c

values = [1, 2, 3]
add_three(values)
```

Fix it using unpacking.

````python
def add_three(a, b, c):
    return a + b + c


values = [1, 2, 3]

print(add_three(*values))
````

### Solution explanation

`add_three(values)` passes one argument: the list itself.

`add_three(*values)` passes three arguments: `1`, `2`, and `3`.

## Problem 16: Build a Flexible Logger

Write a function `log(level, *messages)` that prints a logging level followed by all messages joined with spaces.

Example:

```python
log('INFO', 'User', 'created', 101)
# [INFO] User created 101
```

````python
def log(level, *messages):
    text = ' '.join(str(message) for message in messages)
    print(f'[{level}] {text}')


log('INFO', 'User', 'created', 101)
log('ERROR', 'File not found:', 'data.csv')
````

### Solution explanation

`level` is required.

`*messages` captures any number of message fragments.

Each message is converted to a string before joining, allowing numbers and other objects to be logged safely.

## Problem 18: Function Composition with `*args`

Write a function `compose_two(f, g, *args)` that computes:

```python
f(g(*args))
```

Use it to compute the length of the string produced by joining several words.

````python
def compose_two(f, g, *args):
    return f(g(*args))


def join_words(*words):
    return ' '.join(words)


print(compose_two(len, join_words, 'advanced', 'python', 'args'))
````

### Solution explanation

`compose_two` forwards all extra positional arguments to `g`.

Then it passes the result of `g(*args)` into `f`.

This demonstrates how `*args` supports higher-order functions.

## Problem 19: Avoid Mutable Default Arguments with `*args`

Write a function `collect_items(*items)` that returns a new list containing all supplied items.

Show that each call returns a separate list.

````python
def collect_items(*items):
    return list(items)


a = collect_items(1, 2, 3)
b = collect_items('x', 'y')

a.append(999)

print(a)
print(b)
````

### Solution explanation

`*items` creates a fresh tuple for each function call.

`list(items)` creates a fresh list.

This avoids shared mutable default argument problems.

## Problem 20: Advanced Dispatcher

Create a small command dispatcher.

The function `dispatch(command, *args)` should support:

- `'add'`: return the sum of all arguments
- `'mul'`: return the product of all arguments
- `'max'`: return the maximum argument

Raise `ValueError` for unknown commands.

````python
def dispatch(command, *args):
    if command == 'add':
        return sum(args)

    if command == 'mul':
        result = 1
        for value in args:
            result *= value
        return result

    if command == 'max':
        return max(args)

    raise ValueError(f'Unknown command: {command}')


print(dispatch('add', 1, 2, 3, 4))
print(dispatch('mul', 2, 3, 4))
print(dispatch('max', 10, 7, 99, 3))

try:
    dispatch('min', 1, 2, 3)
except ValueError as ex:
    print(ex)
````

### Solution explanation

`command` is a required positional argument.

`*args` captures the values that belong to the command.

This pattern is useful for interpreters, command-line tools, routing functions, and simple plugin systems.
