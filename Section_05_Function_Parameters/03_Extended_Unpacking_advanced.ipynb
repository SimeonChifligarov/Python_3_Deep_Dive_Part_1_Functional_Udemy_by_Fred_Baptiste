## Problem 1: First, Middle, Last

Given a sequence, write a function `split_edges(seq)` that returns a tuple containing:

- the first element
- a list of the middle elements
- the last element

Example:

```python
split_edges([10, 20, 30, 40, 50])
# (10, [20, 30, 40], 50)
```

````python
def split_edges(seq):
    first, *middle, last = seq
    return first, middle, last


print(split_edges([10, 20, 30, 40, 50]))
print(split_edges('python'))
````

### Solution explanation

`first, *middle, last = seq` assigns the first item to `first`, the last item to `last`, and everything between them to `middle`.

Important detail: the starred target always becomes a list, even when the original iterable is a string or tuple.

## Problem 2: Normalize a Record

A record has this format:

```python
(user_id, username, email, *tags)
```

Write a function `normalize_user(record)` that returns a dictionary:

```python
{
    'id': user_id,
    'username': username,
    'email': email,
    'tags': tags
}
```

````python
def normalize_user(record):
    user_id, username, email, *tags = record
    return {
        'id': user_id,
        'username': username,
        'email': email,
        'tags': tags
    }


record = (101, 'ada', 'ada@example.com', 'admin', 'editor', 'active')
print(normalize_user(record))
````

### Solution explanation

The starred name `tags` collects all remaining values after the required first three fields.

This is useful when records have a fixed prefix and a variable-length tail.

## Problem 3: Nested Unpacking

Given this data:

```python
data = [
    'order-001',
    ('Alice', 'Smith'),
    ['laptop', 'mouse', 'keyboard', 'monitor']
]
```

Use unpacking to extract:

- `order_id`
- `first_name`
- `last_name`
- `main_item`
- `accessories`

Expected result:

```python
order_id = 'order-001'
first_name = 'Alice'
last_name = 'Smith'
main_item = 'laptop'
accessories = ['mouse', 'keyboard', 'monitor']
```

````python
data = [
    'order-001',
    ('Alice', 'Smith'),
    ['laptop', 'mouse', 'keyboard', 'monitor']
]

order_id, (first_name, last_name), (main_item, *accessories) = data

print(order_id)
print(first_name)
print(last_name)
print(main_item)
print(accessories)
````

### Solution explanation

This combines nested unpacking and starred unpacking.

`(first_name, last_name)` unpacks the customer tuple.

`(main_item, *accessories)` unpacks the item list, keeping the first item separate and collecting the rest into a list.

## Problem 5: Merge Configuration Dictionaries

You have three configuration dictionaries:

```python
defaults = {'debug': False, 'timeout': 30, 'retries': 3}
environment = {'timeout': 60}
user_config = {'debug': True}
```

Create a final configuration dictionary where later dictionaries override earlier ones.

````python
defaults = {'debug': False, 'timeout': 30, 'retries': 3}
environment = {'timeout': 60}
user_config = {'debug': True}

final_config = {**defaults, **environment, **user_config}

print(final_config)
````

### Solution explanation

`**` unpacks key-value pairs from dictionaries into a new dictionary.

When keys repeat, the rightmost value wins.

So `environment` overrides `defaults`, and `user_config` overrides both.

## Problem 6: Detect Duplicate Dictionary Keys by Merge Behavior

Given two dictionaries:

```python
left = {'a': 1, 'b': 2, 'c': 3}
right = {'b': 20, 'd': 4}
```

Use dictionary unpacking to merge them, then separately compute which keys were overwritten.

````python
left = {'a': 1, 'b': 2, 'c': 3}
right = {'b': 20, 'd': 4}

merged = {**left, **right}
overwritten_keys = set(left) & set(right)

print(merged)
print(overwritten_keys)
````

### Solution explanation

`{**left, **right}` keeps the value from `right` when a key appears in both dictionaries.

`set(left)` creates a set of keys from `left`, and `set(right)` creates a set of keys from `right`.

The intersection tells us which keys were overwritten.

## Problem 8: Unpack a Generator Safely

Use extended unpacking to extract the first number and the remaining numbers.

Then show that the remaining numbers are stored in a list.

````python
def numbers():
    for n in range(1, 6):
        yield n


first, *rest = numbers()

print(first)
print(rest)
print(type(rest))
````

## Problem 9: Explain the Error

Why does the following code raise an error?

```python
a, *b, *c = [1, 2, 3, 4]
```

````python
# This code is intentionally invalid:
# a, *b, *c = [1, 2, 3, 4]

print('A single unpacking pattern may contain only one starred target.')
````

### Solution explanation

Python allows only one starred target in a single unpacking level.

Otherwise, Python would not know how many values should go into `b` and how many should go into `c`.

Valid example:

```python
a, *b, c = [1, 2, 3, 4]
```

Invalid example:

```python
a, *b, *c = [1, 2, 3, 4]
```

## Problem 10: Multiple Starred Targets in Nested Levels

The following pattern is valid:

```python
a, *b, (c, *d) = [1, 2, 3, [4, 5, 6]]
```

Predict the values of `a`, `b`, `c`, and `d`.

````python
a, *b, (c, *d) = [1, 2, 3, [4, 5, 6]]

print(a)
print(b)
print(c)
print(d)
````

### Solution explanation

At the outer level:

```python
a, *b, last = [1, 2, 3, [4, 5, 6]]
```

So:

```python
a = 1
b = [2, 3]
last = [4, 5, 6]
```

Then the nested pattern unpacks `last`:

```python
c, *d = [4, 5, 6]
```

So:

```python
c = 4
d = [5, 6]
```

## Problem 11: Keep the First Two and Last Two

Write a function `split_two_edges(seq)` that returns:

```python

```

Example:

```python
split_two_edges([10, 20, 30, 40, 50, 60])
# (10, 20, [30, 40], 50, 60)
```

````python
def split_two_edges(seq):
    first, second, *middle, second_last, last = seq
    return first, second, middle, second_last, last


print(split_two_edges([10, 20, 30, 40, 50, 60]))
````

### Solution explanation

The starred target absorbs whatever is left after Python satisfies the non-starred names on both sides.

Here, Python assigns two values from the left, two from the right, and places the remaining values into `middle`.

## Problem 12: Function Argument Unpacking

Given:

```python
def build_profile(username, email, is_admin=False, is_active=True):
    ...
```

Call the function using one positional tuple and one keyword dictionary.

````python
def build_profile(username, email, is_admin=False, is_active=True):
    return {
        'username': username,
        'email': email,
        'is_admin': is_admin,
        'is_active': is_active
    }


args = ('grace', 'grace@example.com')
kwargs = {'is_admin': True}

profile = build_profile(*args, **kwargs)

print(profile)
````

### Solution explanation

`*args` unpacks positional arguments.

`**kwargs` unpacks keyword arguments.

This is commonly used when forwarding arguments from one function to another.

## Problem 14: Convert Starred String Remainder Back to String

Given:

```python
word = 'unpacking'
```

Use unpacking to extract:

- the first character
- the last character
- the middle characters as a string, not a list

````python
word = 'unpacking'

first, *middle_chars, last = word
middle = ''.join(middle_chars)

print(first)
print(middle)
print(last)
````

### Solution explanation

`''.join(middle_chars)` converts that list of characters back into a string.

## Problem 17: Implement a Command Parser

Commands are represented as lists:

```python
['copy', 'file1.txt', 'file2.txt', '--force', '--verbose']
```

The first element is the command name.
The next two elements are required arguments.
The remaining elements are options.

Write `parse_command(parts)` that returns:

```python
{
    'command': command,
    'source': source,
    'destination': destination,
    'options': options
}
```

````python
def parse_command(parts):
    command, source, destination, *options = parts
    return {
        'command': command,
        'source': source,
        'destination': destination,
        'options': options
    }


parts = ['copy', 'file1.txt', 'file2.txt', '--force', '--verbose']
print(parse_command(parts))
````

### Solution explanation

This is a practical use of extended unpacking with command-like structures.

The required positional fields are named directly, and all optional flags are collected into `options`.

## Problem 18: Validate Minimum Length

The following function should require at least three values:

```python
first, *middle, last = seq
```

But that pattern only requires two values.

Write a function that requires at least three values and returns `(first, middle, last)`.

````python
def split_at_least_three(seq):
    first, second, *rest = seq
    *middle, last = [second, *rest]
    return first, middle, last


print(split_at_least_three([1, 2, 3]))
print(split_at_least_three([1, 2, 3, 4, 5]))
````

### Solution explanation

`first, second, *rest = seq` forces the sequence to contain at least two values.

Then `*middle, last = [second, *rest]` forces the remaining part to contain at least one value.

Together, the function requires at least three original values.

## Problem 19: Recursive Head-Tail Processing

Write a recursive function `sum_recursive(seq)` that uses unpacking instead of indexing.

Example:

```python
sum_recursive([1, 2, 3, 4])
# 10
```

````python
def sum_recursive(seq):
    if not seq:
        return 0
    head, *tail = seq
    return head + sum_recursive(tail)


print(sum_recursive([1, 2, 3, 4]))
````

### Solution explanation

`head, *tail = seq` separates the first value from the rest.

This resembles pattern matching in functional programming.

Note: this style is elegant for learning, but for large lists, Python recursion is usually not the best choice.

## Problem 20: Advanced Nested Log Parser

Each log entry has this structure:

```python
entry = (
    '2026-01-01T12:00:00',
    ('api', 'v1', 'users'),
    ('GET', '/users/42', 200, 13.5),
    'cache-hit',
    'authenticated'
)
```

Use unpacking to extract:

- timestamp
- service
- version
- resource
- method
- path
- status_code
- duration_ms
- tags

````python
entry = (
    '2026-01-01T12:00:00',
    ('api', 'v1', 'users'),
    ('GET', '/users/42', 200, 13.5),
    'cache-hit',
    'authenticated'
)

timestamp, (service, version, resource), (method, path, status_code, duration_ms), *tags = entry

print(timestamp)
print(service)
print(version)
print(resource)
print(method)
print(path)
print(status_code)
print(duration_ms)
print(tags)
````

### Solution explanation

This problem combines several concepts:

- fixed-position unpacking
- nested tuple unpacking
- starred collection of remaining values

The final `*tags` captures any number of extra labels after the required fields.
