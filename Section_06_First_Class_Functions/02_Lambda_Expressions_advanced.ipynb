## Problem 1 — Replace Simple Functions with Lambdas

Convert the following simple functions into lambda expressions:

- square a number
- return the last character of a string
- check whether a number is even
- return the larger of two values

Then test each lambda.

````python
square = lambda x: x ** 2
last_char = lambda s: s[-1]
is_even = lambda n: n % 2 == 0
maximum = lambda a, b: a if a > b else b

print(square(5))
print(last_char("lambda"))
print(is_even(10))
print(maximum(7, 3))
````

### Solution Notes

Lambda expressions are best for short, single-expression functions. If the logic becomes multi-step or hard to read, prefer `def`.

## Problem 2 — Lambdas with Defaults, `*args`, and Keyword-Only Arguments

Create a lambda called `build_record` that accepts:

- a required `name`
- any number of positional `scores`
- a keyword-only argument `active`, defaulting to `True`
- arbitrary keyword metadata

It should return a dictionary containing all of that information.

````python
build_record = lambda name, *scores, active=True, **metadata: {
    "name": name,
    "scores": scores,
    "active": active,
    "metadata": metadata
}

record = build_record(
    "Ada",
    95,
    88,
    91,
    active=False,
    course="Python",
    level="advanced"
)

record
````

### Solution Notes

Lambda expressions support most function parameter patterns, including defaults, `*args`, keyword-only arguments, and `**kwargs`. They do not support statement bodies or parameter annotations directly.

## Problem 3 — Build a Generic Function Applicator

Write a function `apply_many(fn, values)` that applies a function to every value in an iterable.

Use lambdas to:

- square every number
- convert every string to lowercase
- extract the length of every string

````python
def apply_many(fn, values):
    """Apply fn to every item in values and return a list."""
    return [fn(value) for value in values]


numbers = [1, 2, 3, 4]
words = ["Python", "Lambda", "Expression"]

print(apply_many(lambda x: x ** 2, numbers))
print(apply_many(lambda s: s.lower(), words))
print(apply_many(lambda s: len(s), words))
````

### Solution Notes

Passing lambdas into higher-order functions is one of their most common uses. However, when the operation already has a named function, such as `str.lower` or `len`, using the named function is usually clearer.

## Problem 4 — Sort Complex Data with Lambda Keys

Given a list of employee dictionaries, sort them by:

1. salary descending
2. department ascending
3. name length ascending

Use lambda expressions as the sorting keys.

````python
employees = [
    {"name": "Ada", "department": "Engineering", "salary": 120000},
    {"name": "Grace", "department": "Research", "salary": 135000},
    {"name": "Linus", "department": "Engineering", "salary": 110000},
    {"name": "Guido", "department": "Language Design", "salary": 125000},
]

by_salary_desc = sorted(employees, key=lambda e: e["salary"], reverse=True)
by_department = sorted(employees, key=lambda e: e["department"])
by_name_length = sorted(employees, key=lambda e: len(e["name"]))

print("Salary descending:")
print(by_salary_desc)

print("\nDepartment ascending:")
print(by_department)

print("\nName length ascending:")
print(by_name_length)
````

### Solution Notes

Lambda functions are very useful as `key` functions for sorting. The key should return the value Python uses for comparison.

## Problem 5 — Multi-Level Sorting with a Lambda

Sort the same employee list by department ascending and then salary descending.

Use a single lambda key.

````python
employees = [
    {"name": "Ada", "department": "Engineering", "salary": 120000},
    {"name": "Grace", "department": "Research", "salary": 135000},
    {"name": "Linus", "department": "Engineering", "salary": 110000},
    {"name": "Guido", "department": "Language Design", "salary": 125000},
]

sorted_employees = sorted(
    employees,
    key=lambda e: (e["department"], -e["salary"])
)

sorted_employees
````

### Solution Notes

Tuple keys allow multi-level sorting. Since strings sort ascending naturally, `department` is used directly. To sort salary descending inside each department, we use `-e['salary']`.

## Problem 6 — Filtering with Lambdas

Given a list of transactions, use lambdas with `filter` to keep only:

- successful transactions
- transactions above 100
- successful transactions above 100

````python
transactions = [
    {"id": 1, "amount": 50, "status": "success"},
    {"id": 2, "amount": 250, "status": "failed"},
    {"id": 3, "amount": 175, "status": "success"},
    {"id": 4, "amount": 20, "status": "success"},
]

successful = list(filter(lambda t: t["status"] == "success", transactions))
large = list(filter(lambda t: t["amount"] > 100, transactions))
successful_large = list(filter(lambda t: t["status"] == "success" and t["amount"] > 100, transactions))

print(successful)
print(large)
print(successful_large)
````

### Solution Notes

`filter` works well with short predicate lambdas. In modern Python, list comprehensions are often more readable, especially when the condition is simple.

## Problem 7 — Map, Filter, and Reduce Pipeline

Given a list of numbers, use lambdas to:

1. keep only positive numbers
2. square them
3. sum the squared values using `functools.reduce`

Return the final result.

````python
from functools import reduce

numbers = [-3, -1, 0, 2, 4, 5]

positive_numbers = filter(lambda x: x > 0, numbers)
squares = map(lambda x: x ** 2, positive_numbers)
total = reduce(lambda a, b: a + b, squares, 0)

total
````

### Solution Notes

This demonstrates functional composition. Still, `sum(x ** 2 for x in numbers if x > 0)` is usually more Pythonic and easier to read.

## Problem 8 — Closures with Lambda Expressions

Write a function `make_multiplier(factor)` that returns a lambda. The returned lambda should multiply its input by `factor`.

Create multipliers for 2, 3, and 10.

````python
def make_multiplier(factor):
    """Return a function that multiplies by factor."""
    return lambda value: value * factor


double = make_multiplier(2)
triple = make_multiplier(3)
times_ten = make_multiplier(10)

print(double(8))
print(triple(8))
print(times_ten(8))
````

### Solution Notes

The lambda remembers the value of `factor` from the enclosing scope. This is called a closure.

## Problem 9 — Late Binding Trap in Lambdas

The following code has a common bug:

```python
funcs = []
for i in range(3):
    funcs.append(lambda x: x + i)
```

Explain the problem and fix it so the lambdas add 0, 1, and 2 respectively.

````python
# Buggy version
funcs_buggy = []
for i in range(3):
    funcs_buggy.append(lambda x: x + i)

print([fn(10) for fn in funcs_buggy])

# Fixed version
funcs_fixed = []
for i in range(3):
    funcs_fixed.append(lambda x, i=i: x + i)

print([fn(10) for fn in funcs_fixed])
````

### Solution Notes

The buggy lambdas all reference the same variable `i`, whose final value after the loop is `2`. The fix captures the current value of `i` as a default argument: `lambda x, i=i: x + i`.

## Problem 10 — Create a Small Command Dispatcher

Build a dictionary called `operations` that maps operation names to lambda expressions:

- `'add'`
- `'subtract'`
- `'multiply'`
- `'divide'`
- `'power'`

Then write a function `calculate(operation, a, b)` that looks up and applies the correct lambda.

````python
operations = {
    "add": lambda a, b: a + b,
    "subtract": lambda a, b: a - b,
    "multiply": lambda a, b: a * b,
    "divide": lambda a, b: a / b,
    "power": lambda a, b: a ** b,
}

def calculate(operation, a, b):
    """Apply a named arithmetic operation to two values."""
    if operation not in operations:
        raise ValueError(f"Unknown operation: {operation}")
    return operations[operation](a, b)


print(calculate("add", 10, 5))
print(calculate("multiply", 10, 5))
print(calculate("power", 2, 8))

try:
    calculate("mod", 10, 3)
except ValueError as exc:
    print(exc)
````

### Solution Notes

A dictionary of functions is often cleaner than a long chain of `if`/`elif` statements. For complex operations, named functions created with `def` are better than lambdas.

## Problem 11 — Compose Functions with Lambdas

Write a function `compose(f, g)` that returns a lambda representing `f(g(x))`.

Use it to create a function that strips whitespace, converts the result to lowercase, and then replaces spaces with hyphens.

````python
def compose(f, g):
    """Return a function equivalent to f(g(x))."""
    return lambda x: f(g(x))


strip_text = lambda s: s.strip()
lower_text = lambda s: s.lower()
hyphenate = lambda s: s.replace(" ", "-")

normalize_then_lower = compose(lower_text, strip_text)
slugify = compose(hyphenate, normalize_then_lower)

slugify("   Advanced Lambda Expressions   ")
````

### Solution Notes

Function composition is common in functional programming. The expression `compose(f, g)` means: first apply `g`, then apply `f` to the result.

## Problem 12 — Introspect Lambda Functions

Create a lambda and inspect its:

- type
- `__name__`
- `__doc__`
- signature

Then manually assign a better `__name__` and `__doc__`.

````python
import inspect

fn = lambda x, y=10: x + y

print(type(fn))
print(fn.__name__)
print(fn.__doc__)
print(inspect.signature(fn))

fn.__name__ = "add_with_default"
fn.__doc__ = "Add x and y, where y defaults to 10."

print("\nAfter metadata update:")
print(fn.__name__)
print(fn.__doc__)
print(inspect.signature(fn))
````

### Solution Notes

Lambda functions are real function objects. Their default name is `'<lambda>'`, which is one reason lambdas are usually not ideal for important reusable logic.
