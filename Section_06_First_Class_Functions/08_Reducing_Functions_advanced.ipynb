# Advanced Problems with Solutions: Reducing Functions in Python

These problems cover `functools.reduce`, custom reductions, initializers, `any`, `all`, products, factorials, aggregation patterns, and when built-ins are better than `reduce`.

## Problem 1: Reimplement `max` Using `reduce`

Use `functools.reduce` to find the maximum value in a non-empty iterable.

```python
from functools import reduce

numbers = [5, 8, 6, 10, 9]

maximum = reduce(lambda a, b: a if a > b else b, numbers)

print(maximum)

# Best practice note:
# In real code, prefer the built-in max(numbers).
# This reduce version is mainly useful for understanding reductions.
```

## Problem 2: Safe Maximum with an Initializer

Write a function `safe_max(sequence, default=None)` that returns the maximum value, or `default` if the sequence is empty.

```python
from functools import reduce

def safe_max(sequence, default=None):
    if not sequence:
        return default
    return reduce(lambda a, b: a if a > b else b, sequence)

print(safe_max([3, 9, 2, 7]))
print(safe_max([], default="No values"))

# Explanation:
# reduce without an initializer cannot reduce an empty iterable.
# We handle the empty case explicitly.
```

## Problem 3: Product of an Iterable

Use `reduce` to calculate the product of all values in a list. Make sure the function works for an empty list too.

```python
from functools import reduce

def product(iterable):
    return reduce(lambda a, b: a * b, iterable, 1)

print(product([2, 3, 4]))
print(product([]))

# 1 is the multiplicative identity.
# That means x * 1 == x, so it is the correct initializer for products.
```

## Problem 4: Factorial with `reduce`

Implement factorial using `reduce`. Handle `0!` correctly.

```python
from functools import reduce

def factorial(n):
    if n < 0:
        raise ValueError("factorial is not defined for negative numbers")
    return reduce(lambda a, b: a * b, range(1, n + 1), 1)

print(factorial(0))
print(factorial(1))
print(factorial(5))

# 0! is 1.
# The initializer 1 ensures this works even when range(1, 1) is empty.
```

## Problem 5: Reimplement `any` with `reduce`

Use `reduce` to determine whether at least one item in a list is truthy.

```python
from functools import reduce

def my_any(iterable):
    return reduce(lambda a, b: bool(a or b), iterable, False)

print(my_any([0, "", None, False]))
print(my_any([0, "", "Python", False]))
print(my_any([]))

# Best practice note:
# Prefer the built-in any(iterable), because it short-circuits.
# reduce does not naturally short-circuit in the same clean way.
```

## Problem 6: Reimplement `all` with `reduce`

Use `reduce` to determine whether every item in a list is truthy.

```python
from functools import reduce

def my_all(iterable):
    return reduce(lambda a, b: bool(a and b), iterable, True)

print(my_all([1, "Python", True]))
print(my_all([1, "", True]))
print(my_all([]))

# all([]) is True because there is no counterexample.
# This is called vacuous truth.
```

## Problem 7: Reduce a List of Dictionaries

Given sales records, use `reduce` to calculate total revenue.

```python
from functools import reduce

sales = [
    {"item": "book", "price": 12.50, "quantity": 3},
    {"item": "pen", "price": 1.20, "quantity": 10},
    {"item": "bag", "price": 35.00, "quantity": 2},
]

total_revenue = reduce(
    lambda total, record: total + record["price"] * record["quantity"],
    sales,
    0
)

print(total_revenue)

# Alternative, usually clearer:
total_revenue_2 = sum(record["price"] * record["quantity"] for record in sales)
print(total_revenue_2)
```

## Problem 8: Build a Frequency Dictionary with `reduce`

Use `reduce` to count how many times each word appears.

```python
from functools import reduce

words = ["python", "java", "python", "c++", "python", "java"]

def add_word_count(counts, word):
    counts[word] = counts.get(word, 0) + 1
    return counts

frequencies = reduce(add_word_count, words, {})

print(frequencies)

# Best practice note:
# For real code, collections.Counter is usually better.

from collections import Counter
print(Counter(words))
```

## Problem 9: Flatten a Nested List

Use `reduce` to flatten a list of lists into a single list.

```python
from functools import reduce

nested = [[1, 2], [3, 4], [], [5, 6]]

flattened = reduce(lambda acc, current: acc + current, nested, [])

print(flattened)

# Better for large data:
# Repeated list concatenation creates many intermediate lists.

flattened_better = [item for sublist in nested for item in sublist]
print(flattened_better)
```

## Problem 10: Running Validation with `all`

Given a list of passwords, check whether all passwords are valid. A valid password has length at least 8 and contains at least one digit.

```python
passwords = ["abc12345", "secure9pass", "hello2024"]

def is_valid_password(password):
    return len(password) >= 8 and any(char.isdigit() for char in password)

all_valid = all(is_valid_password(password) for password in passwords)

print(all_valid)

# This uses two reducing ideas:
# any(...) checks whether at least one character is a digit.
# all(...) checks whether every password is valid.
```

## Problem 11: Find the Longest Word

Use `reduce` to find the longest word. If there is a tie, keep the first longest word.

```python
from functools import reduce

words = ["map", "filter", "reduce", "comprehension", "iterator", "generator"]

longest = reduce(
    lambda current_longest, word: word if len(word) > len(current_longest) else current_longest,
    words
)

print(longest)

# Better built-in solution:
print(max(words, key=len))
```

## Problem 12: Compose Functions with `reduce`

Given a list of functions, use `reduce` to compose them into a single function.

```python
from functools import reduce

def compose(f, g):
    return lambda x: f(g(x))

def double(x):
    return x * 2

def increment(x):
    return x + 1

def square(x):
    return x ** 2

functions = [square, increment, double]

pipeline = reduce(compose, functions)

print(pipeline(3))

# Explanation:
# pipeline(3) = square(increment(double(3)))
# double(3) = 6
# increment(6) = 7
# square(7) = 49
```

## Problem 13: Left Reduction Order Matters

Show that reducing subtraction from left to right gives a different result than grouping from the right.

```python
from functools import reduce

numbers = [10, 3, 2]

left_reduce = reduce(lambda a, b: a - b, numbers)
right_grouped = 10 - (3 - 2)

print(left_reduce)
print(right_grouped)

# reduce works from left to right:
# (10 - 3) - 2 = 5
# This is not the same as 10 - (3 - 2) = 9.
# Reduction order matters for non-associative operations.
```

## Problem 14: Implement a Custom Reduce

Write your own version of `reduce` that supports an optional initializer.

```python
_sentinel = object()

def my_reduce(function, iterable, initializer=_sentinel):
    iterator = iter(iterable)

    if initializer is _sentinel:
        try:
            result = next(iterator)
        except StopIteration:
            raise TypeError("my_reduce() of empty iterable with no initial value")
    else:
        result = initializer

    for item in iterator:
        result = function(result, item)

    return result


print(my_reduce(lambda a, b: a + b, [1, 2, 3]))
print(my_reduce(lambda a, b: a + b, [], 0))
print(my_reduce(lambda a, b: a * b, [2, 3, 4], 1))
```

## Problem 15: Refactor Away from `reduce`

The following `reduce` expression works, but it is not the clearest solution. Refactor it using a built-in function.

```python
from functools import reduce

numbers = [4, 8, 15, 16, 23, 42]

# Works, but not ideal:
total = reduce(lambda a, b: a + b, numbers, 0)

# Better:
better_total = sum(numbers)

print(total)
print(better_total)

# Best practice:
# Use sum, min, max, any, and all when they directly express the goal.
# Use reduce when you truly need a custom accumulation pattern.
```
