# Advanced Problems with Solutions: The `operator` Module

This notebook contains advanced practice problems for Python's `operator` module, including arithmetic operators, comparison operators, `itemgetter`, `attrgetter`, `methodcaller`, sorting, slicing, mutation, reductions, and functional-style programming patterns.

````python
import operator
from functools import reduce, partial
from dataclasses import dataclass
from collections import namedtuple
````

## Problem 1: Replace Lambdas with Operator Functions

Rewrite the following operations using the `operator` module instead of lambdas:

```python
lambda x, y: x + y
lambda x, y: x * y
lambda x, y: x ** y
lambda x, y: x < y
```

````python
# Solution
add = operator.add
multiply = operator.mul
power = operator.pow
less_than = operator.lt

print(add(2, 3))
print(multiply(4, 5))
print(power(2, 4))
print(less_than(10, 20))
````

### Explanation

`operator` functions are named equivalents of common Python operations. They are especially useful with tools like `map`, `filter`, `reduce`, and `sorted`.

## Problem 2: Product with `reduce`

Use `reduce` and `operator.mul` to compute the product of all numbers in a list.

````python
numbers = [2, 3, 4, 5]

# Solution
product = reduce(operator.mul, numbers)
product
````

### Explanation

`operator.mul` replaces `lambda x, y: x * y`, making the reduction more concise and readable.

## Problem 3: Build a Factorial Function

Write a function `factorial(n)` using `range`, `reduce`, and `operator.mul`.

````python
# Solution
def factorial(n):
    if n < 0:
        raise ValueError('factorial is not defined for negative numbers')
    return reduce(operator.mul, range(1, n + 1), 1)

print(factorial(0))
print(factorial(5))
print(factorial(7))
````

### Explanation

The third argument to `reduce` is the initializer. Using `1` handles the `n = 0` case correctly.

## Problem 4: Use `itemgetter` for Sorting Dictionaries

Sort the following users by score descending, then by name ascending.

````python
users = [
    {'name': 'Grace', 'score': 95},
    {'name': 'Ada', 'score': 98},
    {'name': 'Linus', 'score': 95},
    {'name': 'Guido', 'score': 98},
]

# Solution
sorted_users = sorted(users, key=lambda u: (-u['score'], u['name']))
sorted_users
````

### Alternative Using `itemgetter`

`itemgetter` is excellent when the sort direction is the same for all fields. For mixed directions, a small lambda is often clearer.

````python
# Same-direction example
sorted(users, key=operator.itemgetter('score', 'name'))
````

## Problem 5: Extract Multiple Fields with `itemgetter`

Given a list of dictionaries, extract only the `id` and `email` fields from each record.

````python
records = [
    {'id': 1, 'name': 'Ada', 'email': 'ada@example.com'},
    {'id': 2, 'name': 'Grace', 'email': 'grace@example.com'},
    {'id': 3, 'name': 'Linus', 'email': 'linus@example.com'},
]

# Solution
pick_id_email = operator.itemgetter('id', 'email')
list(map(pick_id_email, records))
````

### Explanation

`operator.itemgetter('id', 'email')` returns a callable that extracts both values and returns them as a tuple.

## Problem 6: Sort Objects with `attrgetter`

Create a `Product` class and sort products by price, then by name.

````python
@dataclass
class Product:
    name: str
    price: float
    rating: float

products = [
    Product('Keyboard', 80, 4.7),
    Product('Mouse', 25, 4.5),
    Product('Monitor', 220, 4.8),
    Product('Cable', 25, 4.1),
]

# Solution
sorted(products, key=operator.attrgetter('price', 'name'))
````

### Explanation

`attrgetter('price', 'name')` is equivalent to `lambda obj: (obj.price, obj.name)`.

## Problem 7: Nested Attribute Access

Use `operator.attrgetter` to sort employees by their department name.

````python
@dataclass
class Department:
    name: str

@dataclass
class Employee:
    name: str
    department: Department

employees = [
    Employee('Ada', Department('Research')),
    Employee('Grace', Department('Engineering')),
    Employee('Linus', Department('Infrastructure')),
]

# Solution
sorted(employees, key=operator.attrgetter('department.name'))
````

### Explanation

`attrgetter` supports dotted attribute paths such as `'department.name'`.

## Problem 8: Use `methodcaller` for String Cleaning

Given a list of messy names, strip whitespace and convert them to title case using `operator.methodcaller`.

````python
names = ['  ada lovelace ', ' grace hopper', 'ALAN TURING  ']

# Solution
strip = operator.methodcaller('strip')
title = operator.methodcaller('title')

cleaned = [title(strip(name)) for name in names]
cleaned
````

### Explanation

`methodcaller('strip')` returns a callable equivalent to `lambda obj: obj.strip()`.

## Problem 9: `methodcaller` with Arguments

Use `methodcaller` to replace all hyphens in a string with spaces.

````python
slugs = ['intro-to-python', 'operator-module', 'advanced-functions']

# Solution
replace_hyphen = operator.methodcaller('replace', '-', ' ')
list(map(replace_hyphen, slugs))
````

## Problem 10: Use `operator.truth` with Filtering

Remove falsy values from a mixed list using `filter` and `operator.truth`.

````python
values = [0, 1, '', 'python', [], [1, 2], None, False, True]

# Solution
truthy_values = list(filter(operator.truth, values))
truthy_values
````

## Problem 11: Identity vs Equality

Use `operator.eq`, `operator.is_`, and `operator.is_not` to compare two lists with the same contents.

````python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

# Solution
print(operator.eq(a, b))
print(operator.is_(a, b))
print(operator.is_(a, c))
print(operator.is_not(a, b))
````

### Explanation

`eq` checks value equality. `is_` checks object identity.

## Problem 12: Mutate Lists with `setitem` and `delitem`

Given a list, use `operator.setitem` and `operator.delitem` to modify it in place.

````python
items = ['a', 'b', 'c', 'd']

# Solution
operator.setitem(items, 1, 'B')
operator.delitem(items, 3)
items
````

### Explanation

`setitem(obj, key, value)` performs `obj[key] = value`. `delitem(obj, key)` performs `del obj[key]`.

## Problem 13: Slice with `getitem`, `setitem`, and `delitem`

Use `slice` objects with operator functions to read, replace, and delete ranges in a list.

````python
data = [10, 20, 30, 40, 50, 60]

# Solution
middle = operator.getitem(data, slice(1, 4))
print(middle)

operator.setitem(data, slice(1, 4), ['a', 'b'])
print(data)

operator.delitem(data, slice(2, 4))
print(data)
````

### Explanation

## Problem 14: Count and Locate Items

Use `operator.countOf` and `operator.indexOf` to count and locate elements in a sequence.

````python
letters = ['a', 'b', 'a', 'c', 'a', 'd']

# Solution
print(operator.countOf(letters, 'a'))
print(operator.indexOf(letters, 'c'))
````

## Problem 15: Use `contains` Carefully

Check whether `'py'` appears in `'python'` using `operator.contains`. Then explain the argument order.

````python
# Solution
print(operator.contains('python', 'py'))
print(operator.contains([1, 2, 3], 2))
````

## Problem 16: Build a Reusable Predicate with `partial`

Use `partial` and `operator.contains` to create a function that checks whether an item is in a fixed set of allowed roles.

## Problem 17: Group-Like Sorting with Multiple Getters

Sort transactions by account, then by descending amount. Use `itemgetter` where appropriate.

````python
transactions = [
    {'account': 'A', 'amount': 50},
    {'account': 'B', 'amount': 20},
    {'account': 'A', 'amount': 100},
    {'account': 'B', 'amount': 80},
]

# Solution
get_account = operator.itemgetter('account')
get_amount = operator.itemgetter('amount')

sorted_transactions = sorted(
    transactions,
    key=lambda tx: (get_account(tx), -get_amount(tx))
)

sorted_transactions
````

### Explanation

`itemgetter` extracts fields cleanly. A lambda is still appropriate when extra transformation, such as negating a value, is needed.

## Problem 19: Operator Functions Dispatch to Special Methods

Create a class that implements `__add__`. Show that `operator.add` uses it.

````python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __repr__(self):
        return f'Vector({self.x}, {self.y})'

# Solution
v1 = Vector(1, 2)
v2 = Vector(10, 20)

operator.add(v1, v2)
````

### Explanation

`operator.add(a, b)` calls the same protocol as `a + b`, including custom `__add__` methods.

## Problem 20: In-Place Operators

Compare `operator.add` and `operator.iadd` on a mutable list.

````python
a = [1, 2]
b = [3, 4]

# Solution
result_add = operator.add(a, b)
print('a after add:', a)
print('result_add:', result_add)

result_iadd = operator.iadd(a, b)
print('a after iadd:', a)
print('result_iadd:', result_iadd)
print('same object:', result_iadd is a)
````

### Explanation

`operator.add` creates a new list for list addition. `operator.iadd` performs in-place addition when the object supports it.

## Problem 21: `methodcaller` vs `attrgetter`

Show the difference between retrieving a method and calling a method.

````python
text = 'python'

# Solution
get_upper_method = operator.attrgetter('upper')
call_upper_method = operator.methodcaller('upper')

method = get_upper_method(text)
print(method)
print(method())
print(call_upper_method(text))
````

### Explanation

`attrgetter('upper')` retrieves the method object. `methodcaller('upper')` retrieves and calls it.

## Problem 22: Create a Mini Query Engine

Use `itemgetter`, `attrgetter`, or `methodcaller` to implement reusable transformations over records.

````python
people = [
    {'name': ' ada ', 'age': 36, 'active': True},
    {'name': ' grace ', 'age': 85, 'active': False},
    {'name': ' linus ', 'age': 55, 'active': True},
]

# Solution
get_name = operator.itemgetter('name')
get_active = operator.itemgetter('active')
strip = operator.methodcaller('strip')
title = operator.methodcaller('title')

active_names = [
    title(strip(get_name(person)))
    for person in people
    if get_active(person)
]

active_names
````

### Explanation

Operator callables can make small data-processing pipelines more composable and expressive.

## Problem 23: Build a Safe Dynamic Sort Function

Write a function `sort_by_field(records, field)` that sorts a list of dictionaries by a given field using `operator.itemgetter`. Validate the field name first.

````python
def sort_by_field(records, field):
    if not records:
        return []
    if field not in records[0]:
        raise KeyError(f'Unknown field: {field}')
    return sorted(records, key=operator.itemgetter(field))

# Solution
print(sort_by_field(users, 'name'))
print(sort_by_field(users, 'score'))
````

### Explanation

`itemgetter(field)` is useful when the field to extract is chosen dynamically.

## Problem 24: Boolean Operators Are Not Short-Circuiting

Demonstrate that `operator.and_` and `operator.or_` do not short-circuit like Python's `and` and `or` keywords.

````python
def left():
    print('left called')
    return False

def right():
    print('right called')
    return True

# Keyword short-circuiting
print(left() and right())

# operator.and_ receives already evaluated arguments
print(operator.and_(left(), right()))
````

### Explanation

Function arguments are evaluated before the function call. Therefore, `operator.and_` cannot short-circuit.
