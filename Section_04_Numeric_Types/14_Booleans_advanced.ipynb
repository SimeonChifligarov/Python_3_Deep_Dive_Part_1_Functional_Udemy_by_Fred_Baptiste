## Problem 1: Prove that `bool` is a subclass of `int`

Write code that verifies:

1. `bool` is a subclass of `int`.
2. `True` is an instance of `bool`.
3. `True` is also an instance of `int`.
4. `int(True)` is `1` and `int(False)` is `0`.

````python
print(issubclass(bool, int))
print(isinstance(True, bool))
print(isinstance(True, int))
print(int(True), int(False))
````

### Solution explanation

`bool` inherits from `int`, so `True` and `False` behave like integers in many contexts. Internally, `True` has integer value `1`, while `False` has integer value `0`.

## Problem 2: Equality vs identity

Predict the output of the following expressions, then verify them:

```python
True == 1
True is 1
False == 0
False is 0
True is bool(1)
False is bool(0)
```

````python
print(True == 1)
print(True is 1)
print(False == 0)
print(False is 0)
print(True is bool(1))
print(False is bool(0))
````

### Solution explanation

`==` checks value equality, so `True == 1` and `False == 0` are true. The `is` operator checks object identity. `True` is not the same object as integer `1`, and `False` is not the same object as integer `0`.

## Problem 3: Boolean arithmetic

Use boolean arithmetic to count how many values in a list are positive.

Given:

```python
values = [-3, 0, 4, 10, -1, 8]
```

Count how many values are greater than zero without using an explicit `if` statement.

````python
values = [-3, 0, 4, 10, -1, 8]

positive_count = sum(x > 0 for x in values)

print(positive_count)
````

### Solution explanation

Each comparison `x > 0` produces either `True` or `False`. Since `True` behaves like `1` and `False` behaves like `0`, `sum()` can count how many comparisons are true.

## Problem 4: Chained comparisons

Explain why the following two expressions produce different results:

```python
1 == 2 == False
(1 == 2) == False
```

Then verify the equivalent expanded form of the chained comparison.

````python
expr1 = 1 == 2 == False
expr2 = (1 == 2) == False

expanded = (1 == 2) and (2 == False)

print(expr1)
print(expr2)
print(expanded)
````

### Solution explanation

`1 == 2 == False` is a chained comparison. Python evaluates it like `(1 == 2) and (2 == False)`, which is `False`. But `(1 == 2) == False` first evaluates `(1 == 2)` to `False`, then compares `False == False`, which is `True`.

## Problem 5: Truthiness of built-in containers

Create a table showing the boolean value of the following objects:

```python
0
1
-1
""
"False"
[]
[False]
{}
{0: False}
None
```

````python
objects = [
    0,
    1,
    -1,
    "",
    "False",
    [],
    [False],
    {},
    {0: False},
    None,
]

for obj in objects:
    print(repr(obj).ljust(15), "->", bool(obj))
````

### Solution explanation

Empty containers are falsey. Non-empty containers are truthy, even if their contents are falsey. Therefore, `[False]` and `{0: False}` are both truthy because they are not empty.

## Problem 6: Short-circuit evaluation with `and`

Use short-circuiting to safely check whether the first element of a list is positive.

Your function should return `True` only when the list is non-empty and its first element is greater than zero.

Test it with:

```python
[]
[0]
[5]
[-3]
```

````python
def first_is_positive(items):
    return bool(items and items[0] > 0)

tests = [[], [0], [5], [-3]]

for test in tests:
    print(test, "->", first_is_positive(test))
````

### Solution explanation

`and` short-circuits. If `items` is empty, Python does not evaluate `items[0] > 0`, avoiding an `IndexError`. The final `bool(...)` forces the result to be exactly `True` or `False`.

## Problem 7: Understand what `and` and `or` actually return

Predict and verify the outputs:

```python
"hello" and 123
"" and 123
"hello" or 123
"" or 123
[] or [1, 2, 3]
[0] and "done"
```

````python
expressions = [
    "hello" and 123,
    "" and 123,
    "hello" or 123,
    "" or 123,
    [] or [1, 2, 3],
    [0] and "done",
]

for expression in expressions:
    print(repr(expression))
````

### Solution explanation

`and` and `or` do not always return booleans. They return one of the original operands. `and` returns the first falsey operand or the last operand. `or` returns the first truthy operand or the last operand.

## Problem 8: Avoid the `x == True` anti-pattern

Given the following function:

```python
def is_valid(score):
    return score >= 60
```

Compare these two styles:

```python
if is_valid(score) == True:
    ...

if is_valid(score):
    ...
```

Write a clean version that prints whether each score passes.

````python
def is_valid(score):
    return score >= 60

scores = [95, 60, 59, 0]

for score in scores:
    if is_valid(score):
        print(score, "passes")
    else:
        print(score, "does not pass")
````

### Solution explanation

When a function already returns a boolean, comparing it to `True` is unnecessary. The expression `if is_valid(score):` is clearer and more idiomatic.

users = [

## Problem 10: Count multiple boolean conditions

For each number in the list below, count how many of these conditions are true:

1. The number is positive.
2. The number is even.
3. The number is divisible by 5.

```python
numbers = [-10, -3, 0, 2, 5, 12, 20]
```

````python
numbers = [-10, -3, 0, 2, 5, 12, 20]

for n in numbers:
    count = sum([
        n > 0,
        n % 2 == 0,
        n % 5 == 0,
    ])
    print(n, "->", count)
````

### Solution explanation

Because booleans behave like integers, summing a list of boolean expressions counts how many are true. This is concise and often useful in validation or scoring logic.

## Problem 11: Write a strict boolean validator

Write a function `requires_real_bool(value)` that accepts only actual boolean objects.

It should return the value if `value` is exactly `True` or `False`, but raise `TypeError` for values like `1`, `0`, `"True"`, or `None`.

````python
def requires_real_bool(value):
    if type(value) is not bool:
        raise TypeError("Expected a real bool: True or False")
    return value

tests = [True, False, 1, 0, "True", None]

for value in tests:
    try:
        print(repr(value), "->", requires_real_bool(value))
    except TypeError as error:
        print(repr(value), "->", error)
````

### Solution explanation

`isinstance(1, bool)` is `False`, but `isinstance(True, int)` is `True`. When you need to reject integers and accept only actual booleans, use `type(value) is bool`.

## Problem 12: Detect misleading truthy strings

The string `"False"` is truthy because it is non-empty.

Write a function `parse_bool(text)` that converts only these strings:

- `"true"`, `"yes"`, `"1"` -> `True`
- `"false"`, `"no"`, `"0"` -> `False`

The function should ignore surrounding spaces and letter case. Raise `ValueError` for anything else.

````python
def parse_bool(text):
    normalized = text.strip().lower()
    
    if normalized in {"true", "yes", "1"}:
        return True
    if normalized in {"false", "no", "0"}:
        return False
    
    raise ValueError(f"Cannot parse boolean value from {text!r}")

tests = ["true", "False", " YES ", "0", "maybe", ""]

for text in tests:
    try:
        print(repr(text), "->", parse_bool(text))
    except ValueError as error:
        print(repr(text), "->", error)
````

### Solution explanation

`bool("False")` is `True` because any non-empty string is truthy. To interpret user input as a boolean, parse the string explicitly instead of passing it directly to `bool()`.

## Problem 13: Use `all()` and `any()` correctly

Given a password, validate these rules:

1. At least 8 characters.
2. Contains at least one uppercase letter.
3. Contains at least one lowercase letter.
4. Contains at least one digit.

Return `True` only if all rules pass.

````python
def valid_password(password):
    checks = [
        len(password) >= 8,
        any(ch.isupper() for ch in password),
        any(ch.islower() for ch in password),
        any(ch.isdigit() for ch in password),
    ]
    return all(checks)

passwords = ["abc", "abcdefgh", "Abcdefgh", "Abcdefg1"]

for password in passwords:
    print(password, "->", valid_password(password))
````

### Solution explanation

`any()` checks whether at least one condition in an iterable is true. `all()` checks whether every condition is true. Together, they are ideal for readable validation logic.

## Problem 14: Boolean operator precedence

Predict the result of this expression:

```python
not True or True and False
```

Then rewrite it with parentheses to show Python's actual evaluation order.

````python
expr = not True or True and False
parenthesized = (not True) or (True and False)

print(expr)
print(parenthesized)
````

### Solution explanation

`not` has higher precedence than `and`, and `and` has higher precedence than `or`. Therefore, the expression is evaluated as `(not True) or (True and False)`, which becomes `False or False`, resulting in `False`.

## Problem 15: Build a boolean summary report

Write a function `boolean_report(values)` that returns a dictionary with:

- total number of values
- number of truthy values
- number of falsey values
- whether all values are truthy
- whether any value is truthy

Test it with:

```python
values = [1, 0, "hello", "", [], [1], None, True, False]
```

````python
def boolean_report(values):
    truthy_count = sum(bool(value) for value in values)
    total = len(values)
    
    return {
        "total": total,
        "truthy": truthy_count,
        "falsey": total - truthy_count,
        "all_truthy": all(values),
        "any_truthy": any(values),
    }

values = [1, 0, "hello", "", [], [1], None, True, False]

print(boolean_report(values))
````

### Solution explanation

`bool(value)` converts each value to its truth value. Summing these booleans counts truthy values. `all(values)` checks whether every item is truthy, while `any(values)` checks whether at least one item is truthy.
